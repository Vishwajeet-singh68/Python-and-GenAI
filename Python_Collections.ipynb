**Data Structure and Collections**

```python
def unique_sorted_regions(regions):
    return sorted(set(regions))

# Example
print(unique_sorted_regions(["APAC", "EMEA", "AMER", "APAC"]))
```

```python
def create_price_dict(products, prices):
    return dict(zip(products, prices))

# Example
print(create_price_dict(["Laptop", "Phone", "Tablet"], [70000, 30000, 25000]))
```

```python
coords = (10, 20, 30)

x, y, z = coords
print(f"Coordinates: X={x}, Y={y}, Z={z}")
```

```python
from collections import defaultdict

def group_employees(data):
    result = defaultdict(list)

    for dept, name in data:
        result[dept].append(name)

    return dict(result)

# Example
print(group_employees([("IT", "Alex"), ("HR", "Riya"), ("IT", "John")]))
```

```python
def merge_configs(*configs):
    result = {}
    for config in configs:
        result.update(config)
    return result

# Example
print(merge_configs({"timeout": 10}, {"timeout": 20, "debug": True}))
```

```python
def clean_list(data):
    return [item for item in data if item]

# Example
print(clean_list(["Alex", "", "John", None, "Riya"]))
```

```python
def aggregate_sales(data):
    result = {}

    for category, amount in data:
        result[category] = result.get(category, 0) + amount

    return result

# Example
print(aggregate_sales([("Electronics", 1000), ("Furniture", 2000), ("Electronics", 1500)]))
```

```python
from collections import Counter

def top_n_words(text, n):
    words = text.split()
    counter = Counter(words)
    return counter.most_common(n)

# Example
print(top_n_words("great service great product", 2))
```

```python
def serialize_employees(data):
    return [{"name": name, "department": dept} for name, dept in data]

# Example
print(serialize_employees([("Alex", "IT"), ("Riya", "HR")]))
```
