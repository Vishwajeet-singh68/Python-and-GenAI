**ConditionalAndIterativeStatements**

```python
num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
departments = ["HR", "Finance", "IT", "Sales"]

for dept in departments:
    print(dept)
```

```python
count = 10

while count >= 1:
    print(count)
    count -= 1

print("Blast Off!")
```

```python
production = [120, 95, 110, 80]
target = 100

for i in range(len(production)):
    if production[i] < target:
        print(f"Day {i+1}: {production[i]} units (below target)")
```

```python
attendance = int(input("Enter attendance percentage: "))

if attendance >= 90:
    print("Excellent")
elif 75 <= attendance <= 89:
    print("Satisfactory")
else:
    print("Poor")
```

```python
for i in range(1, 31):
    if i % 2 == 0:
        continue
    if i % 5 == 0:
        continue
    print(i)
```

```python
orders = ["Pending", "Processing", "Delivered", "Failed", "Pending"]

for status in orders:
    if status == "Failed":
        print("Processing stopped due to Failed status.")
        break
    print(f"Processing order: {status}")
```

```python
ages = [16, 21, 18, 25, 14]

count = 0

for age in ages:
    if age > 18:
        count += 1

print("Number of customers above 18:", count)
```

```python
principal = float(input("Enter principal: "))
rate = float(input("Enter rate (%): "))
years = int(input("Enter years: "))

year = 1

while year <= years:
    principal *= (1 + rate / 100)
    print(f"Year {year}: {round(principal, 2)}")
    year += 1
```
