```python
# Temperature conversion
temp_str = input("Enter temperature in Celsius: ")
celsius = float(temp_str)

fahrenheit = (celsius * 9/5) + 32

print("Celsius:", celsius)
print("Fahrenheit:", fahrenheit)
```

```python
# Product price after discount
name = input("Enter product name: ")
price = float(input("Enter price: "))
discount = float(input("Enter discount (%): "))

final_price = price - (price * discount / 100)

print(f"Product: {name}")
print(f"Final Price: {final_price}")
```

```python
# Employee dictionary
employees = {
    101: {"name": "John", "salary": 50000},
    102: {"name": "Alice", "salary": 60000},
    103: {"name": "Bob", "salary": 55000}
}

emp_id = int(input("Enter Employee ID: "))

if emp_id in employees:
    print(employees[emp_id])
else:
    print("Employee not found")
```

```python
# Average sales
sales = [1200, 1500, 1800, 2000, 1700]

avg = sum(sales) / len(sales)

print("Average Sales:", round(avg, 2))
```

```python
# Monthly revenue per branch
revenue = {
    "Delhi": [20000, 25000, 22000],
    "Mumbai": [30000, 28000, 35000],
    "Bangalore": [27000, 26000, 29000]
}

max_branch = None
max_total = 0

for branch, values in revenue.items():
    total = sum(values)
    if total > max_total:
        max_total = total
        max_branch = branch

print("Highest Revenue Branch:", max_branch)
print("Total Revenue:", max_total)
```

```python
# CSV string processing
data = "23,45,67,12"

numbers = list(map(int, data.split(",")))

mean = sum(numbers) / len(numbers)
maximum = max(numbers)

print("Numbers:", numbers)
print("Mean:", mean)
print("Max:", maximum)
```

```python
# Nested dictionary for departments
company = {
    "IT": ["Alice", "Bob"],
    "HR": ["John", "Emma"],
    "Finance": ["David", "Sophia"]
}

dept = input("Enter department: ")

if dept in company:
    print("Employees:", company[dept])
else:
    print("Department not found")
```
