**String Manipulation**

```python
def greet_employee(name, department):
    return f"Welcome {name} to the {department} department!"

# Example
print(greet_employee("John", "Finance"))
```

```python
def format_filename(filename):
    return filename.strip().replace(" ", "_").lower()

# Example
print(format_filename("My Report FINAL .txt"))
```

```python
def normalize_product_ids(ids):
    result = []
    for pid in ids:
        pid = pid.upper().replace("_", "-")
        result.append(pid)
    return result

# Example
print(normalize_product_ids(["prod-001", "PROD_002", "prod-003"]))
```

```python
def parse_url(url):
    parts = url.split("//")
    protocol = parts[0] + "//"
    rest = parts[1].split("/", 1)

    base_url = protocol + rest[0]
    path = "/" + rest[1] if len(rest) > 1 else "/"

    return base_url, path

# Example
base, path = parse_url("https://example.com/products/item1")
print("Base URL:", base)
print("Path:", path)
```

```python
def format_errors(log):
    lines = log.split("\n")
    result = []

    for line in lines:
        if "ERROR" in line:
            line = line.replace("ERROR", "Error")
        elif "WARNING" in line:
            line = line.replace("WARNING", "Warning")

        line = line.replace("at line", "line ")
        line = line.replace(":", " - ")
        result.append(line.strip())

    return "\n".join(result)

# Example
log = "ERROR at line12: missing ; \nWARNING at line15: unused variable"
print(format_errors(log))
```

```python
def clean_csv(text):
    text = text.replace("!", "").replace(",,", ",")
    parts = text.split(",")
    cleaned = [p.strip() for p in parts if p.strip()]
    return ", ".join(cleaned)

# Example
print(clean_csv("John,, Doe!!, New York "))
```

```python
def extract_keywords(text, keywords):
    text_lower = text.lower()
    found = []

    for word in keywords:
        if word.lower() in text_lower:
            found.append(word)

    return found

resume = "John has experience in Python, Django, and React."
print(extract_keywords(resume, ["Python", "Django"]))
```

```python
def generate_email(template, data):
    return template.format(**data)

# Example
template = "Hello {name}, your order {order_id} is confirmed."
data = {"name": "Alice", "order_id": "12345"}

print(generate_email(template, data))
```

```python
def compress_text(text):
    result = ""
    count = 1

    for i in range(1, len(text)):
        if text[i] == text[i-1]:
            count += 1
        else:
            result += text[i-1] + str(count)
            count = 1

    result += text[-1] + str(count)
    return result

# Example
print(compress_text("aaabbcdddd"))
```
