**Function in python**

```python
def calculate_net_salary(gross, tax_rate):
    return gross - (gross * tax_rate / 100)

# Example
print(calculate_net_salary(50000, 10))
```

```python
def format_message(message):
    border = "-" * 40
    print(f"{border}{message}{border}")

# Example
format_message("System Started")
```

```python
def reverse_words(sentence):
    return " ".join(word[::-1] for word in sentence.split())

# Example
print(reverse_words("Hello Python Developer"))
```

```python
def normalize_names(name_list):
    return [name.strip().lower() for name in name_list]

# Example
print(normalize_names([" John ", "MARY", " Alex "]))
```

```python
def load_config(config, key, default_value=None):
    return config.get(key, default_value)

# Example
print(load_config({"timeout": 10}, "timeout"))
```

```python
import time

def measure_time(func):
    start = time.time()
    result = func()
    end = time.time()

    print("Execution time:", (end - start) * 1000, "ms")
    return result

# Example
measure_time(lambda: sum(range(100000)))
```

```python
import os

def scan_directory(path):
    files = []

    for item in os.listdir(path):
        full_path = os.path.join(path, item)

        if os.path.isdir(full_path):
            files.extend(scan_directory(full_path))
        else:
            files.append(full_path)

    return files

# Example (provide valid path)
# print(scan_directory("/project/logs"))
```

```python
def invoke(command_name, commands_dict):
    if command_name in commands_dict:
        return commands_dict[command_name]()
    return "Command not found"

# Example
print(invoke("start", {"start": lambda: "System started"}))
```

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

def generate_fibonacci(n):
    return [fibonacci(i) for i in range(n)]

# Example
print(generate_fibonacci(6))
```
